AAanalysis output tables can be written to a delimited text file together with a JSON metadata sidecar using the ``to_table()`` function. The sidecar documents every exported column, so a table stays interpretable long after the session that produced it. Only ``df`` and ``file_path`` are required:

```python
import json
import tempfile
from pathlib import Path

import pandas as pd
import aaanalysis as aa

aa.options["verbose"] = False
folder_out = Path(tempfile.mkdtemp())

# Load an example CPP feature table
df_feat = aa.load_features()
aa.display_df(df_feat, n_rows=10, show_shape=True)

# Export it together with its metadata sidecar
file_path = str(folder_out / "df_feat.csv")
dict_metadata = aa.to_table(df=df_feat, file_path=file_path)
```

The table is written without the row index, so it reads back with plain pandas and the round-trip is lossless:

```python
df_feat_read = pd.read_csv(file_path)
aa.display_df(df_feat_read, n_rows=10, show_shape=True)
df_feat_read.equals(df_feat)
```

``to_table()`` returns the sidecar record it wrote next to the table as ``<stem>.meta.json``. Its top level states the version of the sidecar layout, the schema the table was documented against, and how the file was written:

```python
{k: v for k, v in dict_metadata.items()
 if k not in ["columns", "columns_undocumented", "categories", "parts", "provenance"]}
```

The ``columns`` entry is the part that makes the export self-describing: every exported column carries its meaning, its contract dtype, the pandas dtype it was written with, and, where the contract defines them, its allowed values or numeric range:

```python
df_columns = pd.DataFrame(dict_metadata["columns"]).T
aa.display_df(df_columns, n_rows=10, show_shape=True)
```

The ``provenance`` entry is the record of ``get_provenance()``: it identifies the exact environment the exported table was produced in, and fingerprints the table itself, so a later run can confirm it starts from the same data:

```python
# The values identify one specific run, so only the recorded keys are shown here
list(dict_metadata["provenance"])
```

Use ``name`` to select the data schema the columns are documented against. It defaults to ``'df_feat'``; any other output table, such as a sequence table, is exported under its own schema. Columns outside the selected contract are still described and listed in ``columns_undocumented``:

```python
df_seq = aa.load_dataset(name="DOM_GSEC", n=5)
dict_metadata_seq = aa.to_table(df=df_seq, file_path=str(folder_out / "df_seq.csv"),
                                name="df_seq")
print(dict_metadata_seq["columns_undocumented"])
aa.display_df(pd.DataFrame(dict_metadata_seq["columns"]).T, n_rows=10, show_shape=True)
```

``sep`` sets the column separator and ``random_state`` records the seed the exported table was computed under. The separator defaults to the one implied by the file extension (a comma for ``.csv``, a tab for ``.tsv``), and a ``.csv`` file also accepts ``';'``:

```python
file_path_tsv = str(folder_out / "df_feat.tsv")
dict_metadata_tsv = aa.to_table(df=df_feat, file_path=file_path_tsv, name="df_feat",
                                sep="\t", random_state=42)
df_feat_tsv = pd.read_csv(file_path_tsv, sep="\t")
aa.display_df(df_feat_tsv, n_rows=10, show_shape=True)
dict_metadata_tsv["provenance"]["random_state"]
```
