You can load all scales from ``AAontology`` (see [Breimann24b]_) and their description using 
the ``load_scales()`` function. The min-max normalized numerical scales are provided by default:

```python
import aaanalysis as aa
aa.options["verbose"] = False
df_scales = aa.load_scales()
print(len(df_scales.T), "scales exist in AAontology")
aa.display_df(df_scales, n_cols=5)
```

To retrieve the un-normalized scales, set ``name='scales_raw'``:

```python
df_scales = aa.load_scales(name="scales_raw")
aa.display_df(df_scales, n_cols=5, n_rows=5)
```

The first 20 PCs of all compressed scales can be retrieved by ``name='scales_pca'``: 


```python
df_pc = aa.load_scales(name="scales_pc")
aa.display_df(df_pc, show_shape=True)
```

The ``AAontolgy`` two-level classification can be retrieved by ``name='scales_cat'``

```python
df_cat = aa.load_scales(name="scales_cat")
aa.display_df(df_cat, n_rows=50, show_shape=True)
```

Additionally, we tested a plethora of scale sets (see [Breimann24a]_) across 12 different protein benchmarking datasets—six at the sequence level and six at the amino acid level, each with three amino acid windows. This resulted in 24 evaluations, for  which we identified the top 60 scale sets, referred to as 'top60'. These can be accessed by setting ``name='top60'``:

```python
df_top60 = aa.load_scales(name="top60")
aa.display_df(df_top60, n_cols=5, show_shape=True)
```

To obtain their evaluation, set ``name='top60_eval'``:

```python
df_top60_eval = aa.load_scales(name="top60_eval")
aa.display_df(df_top60_eval, show_shape=True)
```

You can for example select the scale set with the best performance for the 'SEQ_CAPSID' dataset as follows:

```python
# Sort 'df_top60_eval' by 'SEQ_CAPSID' and get the index of the top row
top_id = df_top60_eval.sort_values(by="SEQ_CAPSID", ascending=False).index[0]
df_scales_top =  aa.load_scales(top60_n=top_id)
# Select a specific set using integers
df_cat_top20 = aa.load_scales(name="scales_cat", top60_n=20)
aa.display_df(df_cat_top20, n_rows=6, show_shape=True)
```

Two optional filtering steps are provided by the ``just_aaindex`` and ``unclassified_out`` parameters:

```python
n_all_scales = len(aa.load_scales().T)
n_just_aaindex = len(aa.load_scales(just_aaindex=True).T)
n_classified = len(aa.load_scales(unclassified_out=True).T)
n_both_filter_steps = len(aa.load_scales(just_aaindex=True, unclassified_out=True).T)
print(n_all_scales, " scales")
print(n_just_aaindex, " from AAindex")
print(n_classified, " are classified in AAontology")
print(n_both_filter_steps, " fulfill both.")
```

For interpretable scale sets, ``top_explain_n`` selects the 5, 10, ..., 60 most interpretable
AAontology subcategories and returns all their scales. Ratings and tiers are available via ``name="subcat"``:

```python
df_cat_explain = aa.load_scales(name="scales_cat", top_explain_n=5)
print(len(aa.load_scales(top_explain_n=5).T), "scales in the 5 most interpretable subcategories")
aa.display_df(df_cat_explain, n_rows=6, show_shape=True)
```

Set ``top_explain_min_th`` to 0.3, 0.4, ..., 0.9 for precomputed ``AAclust`` redundancy reduction:

```python
n_full = len(aa.load_scales(top_explain_n=30).T)
n_reduced = len(aa.load_scales(top_explain_n=30, top_explain_min_th=0.5).T)
print(n_full, "scales in tier 30")
print(n_reduced, "scales after AAclust redundancy reduction (min_th=0.5)")
```

Remaining scale counts across all supported tiers and thresholds:

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

tiers = list(range(5, 65, 5))
thresholds = [i / 10 for i in range(3, 10)]
df_counts = pd.DataFrame(
    [[aa.load_scales(top_explain_n=n, top_explain_min_th=th).shape[1]
      for th in thresholds] for n in tiers],
    index=tiers, columns=thresholds,
)
aa.plot_settings()
fig, ax = plt.subplots(figsize=(7, 5))
sns.heatmap(df_counts, cmap="viridis", annot=True, fmt="d", ax=ax,
            cbar_kws={"label": "Remaining scales"})
ax.set(xlabel="top_explain_min_th", ylabel="top_explain_n")
plt.tight_layout()
plt.show()
```

The first 20 selected subcategories are listed below. Tiers are cumulative; redundancy reduction
may leave some subcategories without a representative scale. Ratings run from 1 (most interpretable) to 10.

```python
df_subcat = aa.load_scales(name="subcat")
df_subcat_top20 = df_subcat.loc[df_subcat["top_explain"] <= 20,
                               ["subcategory", "interpret_grade", "top_explain"]]
aa.display_df(df_subcat_top20, n_rows=20, show_shape=True)
```
