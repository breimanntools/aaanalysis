# `StructurePreprocessor.get_dssp()`

`get_dssp` runs DSSP on each entry's PDB / CIF file and returns a copy of `df_seq` with the **raw**, inspectable per-residue DSSP output (one list per residue column: `ss`, `asa`, `phi`, `psi`, hydrogen-bond offsets and energies) plus a `dssp_ok` flag. It is the curatable intermediate that `encode_dssp` turns into a `dict_num`. The chain whose ATOM sequence best matches `df_seq['sequence']` is selected automatically and aligned back to the full sequence.

Requires `aaanalysis[pro]` plus a `mkdssp` (or legacy `dssp`) binary on `PATH`. The structures are AlphaFold models fetched with `fetch_alphafold`.

```python
import warnings
import tempfile
from pathlib import Path
import aaanalysis as aa
aa.options['verbose'] = False
warnings.filterwarnings('ignore')

# Three short human proteins from the bundled gamma-secretase set; their ``entry``
# values are UniProt accessions, which is what AlphaFold DB is keyed by.
df_seq = aa.load_dataset(name='DOM_GSEC', n=10)
df_seq = df_seq[df_seq['entry'].isin(['Q14802', 'O43914', 'P01135'])].reset_index(drop=True)

strp = aa.StructurePreprocessor(verbose=False)
af_dir = Path(tempfile.mkdtemp()) / 'alphafold'   # use a persistent project folder in real work

strp.fetch_alphafold(df_seq=df_seq, out_folder=af_dir)

df_dssp = strp.get_dssp(df_seq=df_seq, pdb_folder=af_dir)
print('added columns:', [c for c in df_dssp.columns if c not in df_seq.columns])
aa.display_df(df_dssp[['entry', 'gene', 'ss', 'asa', 'dssp_ok']], n_rows=10, show_shape=True, char_limit=40)
```

Each list column has one element per residue of the full sequence, so the secondary-structure string lines up with the sequence:

```python
for entry, seq, ss in zip(df_dssp['entry'], df_dssp['sequence'], df_dssp['ss']):
    print(entry)
    print('  seq', seq[:60])
    print('  ss ', ''.join(ss[:60]))
```

## Further parameters

`features` restricts the DSSP columns that are added (any subset of `'ss'`, `'asa'`, `'phi_psi'`, `'hbonds'`); `ss_mode` chooses the three-state (`'ss3'`: helix / strand / coil) or the native eight-state (`'ss8'`) alphabet; `gap_handling` controls residues DSSP skips: `'pad'` keeps the full sequence length with placeholders, `'omit'` drops them so the lists are shorter than the sequence.

```python
df_dssp8 = strp.get_dssp(df_seq=df_seq, pdb_folder=af_dir, features=['ss', 'asa'],
                         ss_mode='ss8', gap_handling='omit')
print('eight-state codes:', ''.join(df_dssp8['ss'].iloc[0][:60]))
print('list length vs sequence length:', len(df_dssp8['ss'].iloc[0]), 'vs', len(df_dssp8['sequence'].iloc[0]))
aa.display_df(df_dssp8[['entry', 'ss', 'asa', 'dssp_ok']], n_rows=10, show_shape=True, char_limit=40)
```
