After :meth:`CPPGrid.run`, :meth:`CPPGrid.eval` scores every configuration and returns ``df_params`` joined to per-configuration quality columns (``avg_ABS_AUC`` is the mean of each feature table's ``abs_auc``), sorted best-first so ``df.iloc[0]`` is the best configuration.

```python
import aaanalysis as aa
aa.options["verbose"] = False

df_seq = aa.load_dataset(name="DOM_GSEC", n=10)
labels = df_seq["label"].to_list()
cppg = aa.CPPGrid(df_seq=df_seq, labels=labels, n_jobs=1, random_state=0)
cppg.run(params_cpp={"n_filter": [10, 25]})
df_eval = cppg.eval(sort_by="avg_ABS_AUC")
aa.display_df(df_eval, n_rows=10, show_shape=True)
```

The best configuration's feature table is then ``grid.list_df_feat_[df_eval.index[0]]``.

```python
best_i = df_eval.index[0]
aa.display_df(cppg.list_df_feat_[best_i], n_rows=10, show_shape=True)
```

**Further parameters.** ``CPPGrid.eval`` also accepts: ``ascending`` — Sort direction.

```python
# Further parameters: ascending flips the sort direction (None picks the sensible default,
# descending for higher-is-better metrics such as avg_ABS_AUC)
df_eval = cppg.eval(sort_by="avg_ABS_AUC", ascending=True)
aa.display_df(df_eval, n_rows=10, show_shape=True)
```
